```python
import pandas as pd

CALC_TYPES = [
    "mbar",
    "dg-c2-pb",
    "dg-c2-gb",
    "dg-ie-pb",
    "dg-ie-gb",
    "dh-pb",
    "dh-gb",
]


# Define the custom orders
SYSTEM_ORDER = ["p38", "A2A", "ptp1b", "tyk2", "thrombin", "mcl1", "CyclophilinD", "SAMPL6-OA"]
FORCE_FIELD_ORDER = ['espaloma-0.3.1', 'gaff-2.11', 'openff-2.0.0']

SYSTEM_NAME = {
    "p38": "P38", 
    "A2A": "A2A",
    "ptp1b": "PTP1B",
    "tyk2": "TYK2",
    "thrombin": "Thrombin",
    "mcl1": "MCL1",
    "CyclophilinD": "CyclophilinD",
    "SAMPL6-OA": "SAMPL6-OA"
}


BindFlowData = pd.read_csv("../BindFlow.csv", index_col=0)

columns = [
    "system",
    "ligand",
    "replica",
    "sample",
    "exp_dG",
    "exp_dG_error",
]
for CALC_TYPE in CALC_TYPES:
    columns += [
    f"simulation_{CALC_TYPE}_espaloma-0.3.1",
    f"simulation_{CALC_TYPE}_gaff-2.11",
    f"simulation_{CALC_TYPE}_openff-2.0.0",
    ]
BindFlowData = BindFlowData[columns]

BindFlowData.rename(
    columns={
        "system": "source",
    },
    inplace=True
)
```

```python

SIMU = {
    "fep" : ["mbar"],
    "mm(pb/gb)sa": [
        "dg-c2-pb",
        "dg-c2-gb",
        "dg-ie-pb",
        "dg-ie-gb",
        "dh-pb",
        "dh-gb",
    ]
}

for key, items in SIMU.items():
    count = 0
    for item in items:
        for ff in FORCE_FIELD_ORDER:
            count += sum(BindFlowData[f"simulation_{item}_{ff}"].notna())
    if key == "fep":
        print(f"{key}: {count}")
    else:
        print(f"{key}: {count/20}")
```
